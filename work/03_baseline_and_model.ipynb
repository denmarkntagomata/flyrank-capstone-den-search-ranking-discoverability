This notebook trains the Random Forest momentum classifier and compares it against a simple rule-based baseline on the out-of-time validation window defined in notebook 02.

## Takeaways
- The baseline (position-slope sign only) reaches ~47% accuracy, only slightly better than random three-way guessing (~33%) — position slope alone is a weak signal on its own.
- The Random Forest, using position, CTR-gap, and impression/click trend features together, reaches ~94% accuracy and ~0.94 macro-F1 on the held-out, strictly-later validation window — a large, meaningful lift over the baseline on the *same* split.
